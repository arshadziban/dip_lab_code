```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Load image in grayscale
image = cv2.imread('images\image.png', 0)

# Define a kernel for convolution and correlation
kernel = np.array([[1, 0, -1],
                   [1, 0, -1],
                   [1, 0, -1]])

# Apply correlation (filter2D performs correlation)
correlation = cv2.filter2D(image, -1, kernel)

# Apply convolution (flip kernel before applying)
flipped_kernel = np.flip(kernel)
convolution = cv2.filter2D(image, -1, flipped_kernel)

# Apply Gaussian blur
blur = cv2.GaussianBlur(image, (5, 5), 0)

# Edge detection using Canny
edges = cv2.Canny(image, 100, 200)

# Display results
plt.figure(figsize=(12, 6))

plt.subplot(2, 3, 1)
plt.imshow(image, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(correlation, cmap='gray')
plt.title("Correlation")
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(convolution, cmap='gray')
plt.title("Convolution")
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(blur, cmap='gray')
plt.title("Gaussian Blur")
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(edges, cmap='gray')
plt.title("Edge Detection")
plt.axis('off')

plt.tight_layout()
plt.show()
```
